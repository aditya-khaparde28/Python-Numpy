```python
#Condition based choices
#Where
```

```python
arr=np.array([1,2,3])
result=np.where(arr<2,"low","high")
```

```python
print(result)
```

```python
#ArgWhere gives row column position for given array where the condition is staisfied
print(np.argwhere(arr>2))
```

```python
mask=np.logical_and(arr>2,arr<5)
print(mask)
```

```python
print(np.argwhere(arr>3))
```

```python
#Masking Array
mask=np.logical_or(arr1>2,arr1>9)
print(mask)
```

```python
arr1[mask]
```
